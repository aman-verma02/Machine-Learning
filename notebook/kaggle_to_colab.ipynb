```python
import os
from google.colab import userdata

# Retrieve Kaggle credentials from Colab Secrets
try:
    os.environ['KAGGLE_USERNAME'] = userdata.get('KAGGLE_USERNAME')
    os.environ['KAGGLE_KEY'] = userdata.get('KAGGLE_KEY')
    print("Kaggle credentials successfully loaded from Colab Secrets!")
except Exception as e:
    print("Error loading credentials. Please ensure you have added 'KAGGLE_USERNAME' and 'KAGGLE_KEY' to your Colab Secrets (🔑 icon on the left panel).")
    print(e)
```
